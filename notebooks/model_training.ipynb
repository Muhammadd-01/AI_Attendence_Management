# Model Training and Enrollment

We use a pre-trained ResNet model via the `face_recognition` library (dlib backend). 
Therefore, "training" in this context refers to **enrollment** (generating embeddings for our dataset).

## 1. The Model
The 128-dimensional embedding model is trained on a dataset of 3 million faces and achieves 99.38% accuracy on the Labeled Faces in the Wild (LFW) benchmark.

## 2. Enrollment Process
1. Loop through all 100 images for a student.
2. Detect the face in each image.
3. Pass the face through the ResNet model to get a 128-d vector.
4. Store these vectors in Firebase Firestore under `students/{id}/encodings`.

## 3. Real-time Matching (Inference)
At inference time, the camera captures a face, generates its 128-d embedding, and calculates the Euclidean distance to all stored embeddings.
- If `distance < 0.55`, it is a match.
- `Confidence = 1.0 - distance`
