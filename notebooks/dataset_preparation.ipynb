# Dataset Preparation

This notebook documents the data collection and preprocessing methodology for the Smart Attendance System.

## 1. Face Capture
Students are registered via the UI. 100 images are captured at various angles (straight, left, right, up, down) and expressions. They are stored in `backend/dataset/{student_id}/`.

## 2. Preprocessing
- **Lighting Normalization:** Histogram equalization is applied to the LAB L-channel.
- **Blur Check:** Laplacian variance is used. Images with variance < 50.0 are rejected.
- **Resizing:** Faces are cropped and resized to 150x150 pixels for uniform processing.

## 3. Dataset Requirements
- Target: 6-8 students
- Images: ~100 per student (600-800 total)
- Resolution: Varies by camera, cropped tightly to face bounds.
