# Model Testing & Evaluation

This notebook evaluates the face recognition pipeline against the SRS requirements.

## Targets
- Recognition Time: ≤ 2 seconds
- Accuracy: ≥ 95%

## Methodology
1. **Hold-out Test Set:** 20 images per student are kept aside during enrollment.
2. **Negative Test Set:** Images of people NOT in the dataset.
3. **Metrics:**
   - True Positives (TP): Registered student correctly identified.
   - True Negatives (TN): Unknown person correctly labeled as Unknown.
   - False Positives (FP): Unknown person incorrectly labeled as a student (Critical failure/Proxy).
   - False Negatives (FN): Registered student labeled as Unknown.

## Threshold Tuning
- The default `FACE_MATCH_THRESHOLD` is `0.55`.
- A lower threshold (e.g., `0.4`) reduces False Positives (better security) but increases False Negatives.
- A higher threshold (e.g., `0.6`) reduces False Negatives but risks False Positives (proxy attendance).
- The UI allows dynamic tuning of this threshold in Settings.
